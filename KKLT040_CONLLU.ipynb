### 2.1 Most frequent tokens + lemmas in a text

Question: What changes when you look at most frequent lemmas instead of tokens?

Question: What about when you filter the text from some part-of-speech classes?


```python
from analyze import most_frequent
print("Most frequent tokens")
print(most_frequent("data/sr.conllu", "FORM", 15)) # the number defines how many we want to see

print("Most frequent lemmas")
print(most_frequent("data/sr.conllu", "LEMMA", 15)) # the number defines how many we want to see
```

```python
from analyze import most_frequent_filter2
print("Most frequent tokens when filtered")
print(most_frequent_filter2("data/sr.conllu", "FORM", 15)) # the number defines how many we want to see

print("Most frequent lemmas when filtered")
print(most_frequent_filter2("data/sr.conllu", "LEMMA", 15)) # the number defines how many we want to see
```
